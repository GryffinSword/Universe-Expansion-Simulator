```python
# Import libraries
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd # testing
```

```python
# read data in
df = pd.read_csv("table_objsearch_div.csv")
```

```python
# Create figure/plot (will delete later)
plt.scatter(df["RA"], df["Dec"], s=2, alpha=0.5, c=df['Redshift (z)'], cmap='hot')
plt.grid(True)
plt.ylabel("Right Ascension (RA)")
plt.xlabel("Declination (Dec)")
plt.show()
```

```python
# make arrays
RA = df["RA"]
Dec = df["Dec"]

# change to radians
RA_rad = np.radians(RA -180) # centers the point bunches
Dec_rad = np.radians(Dec)

# create the plot
fig = plt.figure(figsize=(10,6))
ax = plt.axes(projection="mollweide")
ax.scatter(RA_rad, Dec_rad, s=2, alpha=0.6, c=df['Redshift (z)'], cmap='hot')
ax.grid(True)
plt.title("Plotting Galaxies Using Latitude/Longitude Map")
plt.show()
```
